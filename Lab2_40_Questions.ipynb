```python
#Concatenate Along Rows (dim=0)
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

c = torch.cat((a, b), dim=0)

print(c)
```

```python
#Concatenate Along Columns (dim=1)
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

c = torch.cat((a, b), dim=1)

print(c)
```

```python
#21. Sum of each row and each column in a 2x3 tensor
t = torch.tensor([[1, 2, 3], [4, 5, 6]])
row_sums = t.sum(dim=1)
column_sums = t.sum(dim=0)
print(f"Sum of each row: {row_sums}")
print(f"Sum of each column: {column_sums}")
```

```python
# 27. Add an extra dimension to make the shape (1, 5)
t = torch.tensor([1, 2, 3, 4, 5])
t_with_extra_dimension = t.unsqueeze(0)
print(t_with_extra_dimension)
print(t_with_extra_dimension.shape)
```

```python
# 29. Clone a tensor and confirm the clone is unaffected by changes to the original
original = torch.tensor([1, 2, 3])
clone = original.clone()
original[0] = 99
print(f"Original: {original}")
print(f"Clone: {clone}")
```

```python
# 30. Assign a tensor to another variable and confirm both refer to the same data
original = torch.tensor([1, 2, 3])
new_variable = original
original[0] = 99
print(f"Original: {original}")
print(f"New variable: {new_variable}")
```
